### Converter GeoDataFrame para Pandas DataFrame

Para converter o `gdf` (GeoDataFrame) em um DataFrame Pandas padrão, precisamos lidar com a coluna `geometry`. Transformaremos os objetos `geometry` em sua representação Well-Known Text (WKT), que é uma maneira comum de armazenar dados espaciais em formato de texto dentro de um DataFrame regular. Posteriormente, a coluna `geometry` original será removida.

```python
import pandas as pd

# Cria uma cópia do GeoDataFrame para evitar modificar o original
df_pandas_distrito = gdf_distrito.copy()

# Converte a coluna 'geometry' para o formato Well-Known Text (WKT)
df_pandas_distrito['geometry_wkt'] = df_pandas_distrito['geometry'].apply(lambda geom: geom.wkt if geom else None)

# Remove a coluna 'geometry' original, pois não é um tipo padrão do pandas
df_pandas_distrito = df_pandas_distrito.drop(columns=['geometry'])

print("Cabeçalho do DataFrame Pandas (com geometria como WKT):")
display(df_pandas_distrito.head())

print("\nInformações do DataFrame Pandas:")
df_pandas_distrito.info()
```

```python
from unidecode import unidecode
import re

def remove_accents_and_normalize(text, keep_spaces=False):
    if isinstance(text, str):
        text = unidecode(text).upper()
        # Substitui separadores comuns como '/', '.', ',', etc., por espaços.
        text = re.sub(r'[^\w\s-]', ' ', text) # Mantém alfanuméricos, espaços e hifens

        # Normaliza espaços - substitui múltiplos espaços por um único espaço
        text = re.sub(r'\s+', ' ', text).strip()

        if not keep_spaces:
            # Substitui todos os espaços por hifens, mas apenas se keep_spaces for False
            text = re.sub(r'\s', '-', text)

            # Normaliza hifens - substitui múltiplos hifens por um único hifen
            # e remove hifens iniciais/finais.
            text = re.sub(r'-+', '-', text).strip('-')

        return text
    return text
```

```python
# Cria uma cópia do GeoDataFrame para evitar modificar o original
df_pandas_subprefeitura = gdf_subprefeitura.copy()

# Converte a coluna 'geometry' para o formato Well-Known Text (WKT)
df_pandas_subprefeitura['geometry_wkt'] = df_pandas_subprefeitura['geometry'].apply(lambda geom: geom.wkt if geom else None)

# Remove a coluna 'geometry' original, pois não é um tipo padrão do pandas
df_pandas_subprefeitura = df_pandas_subprefeitura.drop(columns=['geometry'])

print("Cabeçalho do DataFrame Pandas (com geometria como WKT):")
display(df_pandas_subprefeitura.head())

print("\nInformações do DataFrame Pandas:")
df_pandas_subprefeitura.info()
```

```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
import numpy as np

# Inicializa o geocodificador Nominatim com um timeout maior
geolocator = Nominatim(user_agent="geo_flooding_analysis", timeout=1)

# Adiciona um limitador de taxa para evitar bloqueios da API (1 requisição por segundo)
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)

print("Geocodificador Nominatim configurado com timeout de 1 segundos.")
```

### Refinando a Coluna 'distrito' com Geocodificação

```python
def populate_distrito_column_from_subprefeitura_refined(row):
    subprefeitura_name = row['subprefeitura']
    logradouro = row['logradouro']

    manual_mapping = {
        'ARICANDUVA-VILA-FORMOSA': 'ARICANDUVA-FORMOSA-CARRAO',
        'FREGUESIA-DO-O': 'FREGUESIA-BRASILANDIA'
    }
    subprefeitura_name_for_lookup = manual_mapping.get(subprefeitura_name, subprefeitura_name)

    distritos_found = encontrar_distrito_pela_subprefeitura(subprefeitura_name_for_lookup)

    if not distritos_found:
        # print(f"Nenhum distrito encontrado para subprefeitura '{subprefeitura_name}'")
        return None # Nenhum distrito encontrado para a subprefeitura

    if len(distritos_found) == 1:
        # print(f"Distrito único encontrado para '{subprefeitura_name}': {distritos_found[0]}")
        return distritos_found[0] # Se apenas um distrito, usa-o diretamente

    # Se múltiplos distritos, tenta refinar usando o logradouro
    best_distrito = None
    best_match_score = -1 # Score mais alto indica melhor correspondência

    for distrito_name in distritos_found:
        # Cria uma query para o logradouro e distrito específicos
        query = f"{logradouro}, {distrito_name}, São Paulo, Brazil"
        try:
            location = geocode(query)

            if location:
                # Verifica se o logradouro e o distrito fazem parte do endereço geocodificado
                address_lower = location.address.lower()
                logradouro_lower = logradouro.lower()
                distrito_lower = distrito_name.lower()

                current_score = 0
                if logradouro_lower in address_lower:
                    current_score += 1
                if distrito_lower in address_lower: # Dá maior preferência se o distrito estiver claro no endereço
                    current_score += 2

                if current_score > best_match_score:
                    best_match_score = current_score
                    best_distrito = distrito_name

        except Exception as e:
            # print(f"Erro ao geocodificar '{query}' para refinamento: {e}")
            pass # Ignora erros nas tentativas de refinamento

    if best_distrito:
        print(f"Refinado para '{best_distrito}' para logradouro '{logradouro}' na subprefeitura '{subprefeitura_name}'")
        return best_distrito
    elif distritos_found:
        print(f"Não foi possível refinar para um distrito único, usando o primeiro: '{distritos_found[0]}' para '{logradouro}' na subprefeitura '{subprefeitura_name}'")
        # Se nenhuma correspondência específica por geocodificação foi forte, retorna o primeiro distrito como fallback
        return distritos_found[0]
    else:
        # print(f"Nenhum distrito final determinado para '{logradouro}' na subprefeitura '{subprefeitura_name}'")
        return None # Não deve acontecer se distritos_found não estiver vazio

# Aplica a função refinada ao DataFrame df_alagamentos_cge
df_alagamentos_cge['distrito'] = df_alagamentos_cge.apply(populate_distrito_column_from_subprefeitura_refined, axis=1)

print("DataFrame 'df_alagamentos_cge' após preencher e refinar a coluna 'distrito':")
display(df_alagamentos_cge.head())
```

### Gerando Colunas de Latitude e Longitude


```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

# Re-inicializa o geocodificador Nominatim com um timeout de 5 segundos para coordenadas
geolocator = Nominatim(user_agent="geo_flooding_analysis", timeout=5)

# Adiciona um limitador de taxa para evitar bloqueios da API (1 requisição por segundo)
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)

print("Geocodificador Nominatim reconfigurado com timeout de 5 segundos para latitude/longitude.")
```

```python
def get_coordinates_from_address(row):
    logradouro = row['logradouro']
    referencia = row['referencia']
    distrito = row['distrito'] # 'distrito' deve ser um único distrito ou None

    # Prioriza o primeiro distrito se houver múltiplos separados por vírgula
    single_distrito = None
    if isinstance(distrito, str) and distrito:
        single_distrito = distrito.split(',')[0].strip()

    # Tenta geocodificar apenas pelo distrito primeiro para maior chance de sucesso
    if single_distrito:
        query_distrito_only = f"{single_distrito}, São Paulo, Brazil"
        try:
            location_distrito = geocode(query_distrito_only)
            if location_distrito:
                print(f"Coordenadas encontradas (apenas distrito) para '{query_distrito_only}': ({location_distrito.latitude}, {location_distrito.longitude})")
                return location_distrito.latitude, location_distrito.longitude
            else:
                print(f"Não foi possível encontrar coordenadas (apenas distrito) para '{query_distrito_only}'")
        except Exception as e:
            print(f"Erro ao geocodificar (apenas distrito) '{query_distrito_only}': {e}")

    # Se falhou com apenas o distrito, tenta a query mais específica (logradouro + referencia + distrito)
    address_parts_full = [logradouro]
    if referencia != 0:
        address_parts_full.append(str(referencia))
    if single_distrito:
        address_parts_full.append(single_distrito)
    address_parts_full.append('São Paulo')
    address_parts_full.append('Brazil')

    query_full = ', '.join(address_parts_full)

    try:
        location_full = geocode(query_full)
        if location_full:
            print(f"Coordenadas encontradas (completa) para '{query_full}': ({location_full.latitude}, {location_full.longitude})")
            return location_full.latitude, location_full.longitude
        else:
            print(f"Não foi possível encontrar coordenadas (completa) para '{query_full}'")
            return np.nan, np.nan
    except Exception as e:
        print(f"Erro ao geocodificar (completa) '{query_full}': {e}")
        return np.nan, np.nan

# Aplica a função para criar as colunas de latitude e longitude
df_alagamentos_cge[['latitude', 'longitude']] = df_alagamentos_cge.apply(lambda row: pd.Series(get_coordinates_from_address(row)), axis=1)

print("DataFrame 'df_alagamentos_cge' após adicionar as colunas de latitude e longitude:")
display(df_alagamentos_cge.head())
print("Tipos de dados das novas colunas:")
print(df_alagamentos_cge[['latitude', 'longitude']].dtypes)
```

```python
# Converte a coluna 'data' para o tipo datetime, se ainda não estiver
df_alagamentos_cge['data'] = pd.to_datetime(df_alagamentos_cge['data'])

# Combina 'data' com 'hora_inicio' para criar 'dt_hr_inicio'
df_alagamentos_cge['dt_hr_inicio'] = df_alagamentos_cge.apply(lambda row: pd.to_datetime(f"{row['data'].date()} {row['hora_inicio']}"), axis=1)

# Combina 'data' com 'hora_fim' para criar 'dt_hr_fim'
df_alagamentos_cge['dt_hr_fim'] = df_alagamentos_cge.apply(lambda row: pd.to_datetime(f"{row['data'].date()} {row['hora_fim']}"), axis=1)

# Reordena as colunas para colocar as novas colunas ao lado de 'data'
cols = df_alagamentos_cge.columns.tolist()
insert_idx = cols.index('data') + 1
cols.insert(insert_idx, 'dt_hr_inicio')
cols.insert(insert_idx + 1, 'dt_hr_fim')
cols.remove('dt_hr_inicio') # Remove as duplicatas antes de reatribuir
cols.remove('dt_hr_fim') # Remove as duplicatas antes de reatribuir

df_alagamentos_cge = df_alagamentos_cge[cols]

print("DataFrame 'df_alagamentos_cge' após adicionar as colunas 'dt_hr_inicio' e 'dt_hr_fim':")
display(df_alagamentos_cge.head())
print("Tipos de dados das novas colunas:")
print(df_alagamentos_cge[['dt_hr_inicio', 'dt_hr_fim']].dtypes)
```
